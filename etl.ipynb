```python
import pandas as pd

# Load only first 10000 rows
df = pd.read_csv("complaints_raw.csv.zip", nrows=10000)

# Remove duplicates
df = df.drop_duplicates()

# Convert dates
df["Date received"] = pd.to_datetime(df["Date received"])

# Create date columns
df["year"] = df["Date received"].dt.year
df["month"] = df["Date received"].dt.month

# Save cleaned file
df.to_csv("complaints_cleaned.csv", index=False)

print(df.head())

print("Done")
```
